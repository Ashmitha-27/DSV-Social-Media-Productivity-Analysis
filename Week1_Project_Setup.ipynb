## Problem Statement

Excessive use of social media can affect productivity, sleep, stress levels
and overall digital well-being.

However, the relationship between social media usage and these factors
is not always clearly understood.

This project aims to analyse social media usage patterns and study how they
relate to productivity and well-being using data science and visualization.

## Project Objectives

1. Analyse social media usage patterns.
2. Study the relationship between social media usage and productivity.
3. Analyse social media usage with sleep and stress levels.
4. Identify patterns related to burnout and digital well-being.

6. Detect excessive or unusual usage patterns.
7. Visualize important relationships using suitable charts.
8. Derive meaningful insights and recommendations.
