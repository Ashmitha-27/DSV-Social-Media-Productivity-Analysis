## Missing Value Inspection

Missing values are identified to understand which attributes contain
incomplete data before the cleaning process.

```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:\n")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage (%)": missing_percentage.round(2)
})

missing_report
```

### Missing Value Visualization

```python
missing = df.isnull().sum()

missing.plot(kind='bar')

plt.title('Missing Values in Each Column')
plt.xlabel('Columns')
plt.ylabel('Number of Missing Values')
plt.xticks(rotation=95)
plt.tight_layout()
plt.show()
```

## Duplicate Record Check

Duplicate records are checked to identify repeated observations that may
affect the accuracy of the analysis.

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

### Observation

The dataset contains no duplicate records. Therefore, duplicate removal
is not required at this stage.

## First Impression Report

The First Impression Report summarizes the initial condition of the dataset
before any cleaning is performed.

It includes dataset size, data types, missing values, duplicate records
and basic statistical information.

```python
print("========== FIRST IMPRESSION REPORT ==========")

print("\nDataset Shape:")
print(df.shape)

print("\nTotal Number of Missing Values:")
print(df.isnull().sum().sum())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

print("\nData Type Distribution:")
print(df.dtypes.value_counts())

print("\nBasic Statistical Summary:")
display(df.describe())
```
